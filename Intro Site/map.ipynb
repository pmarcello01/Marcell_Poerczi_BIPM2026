```python
m = folium.Map(location=[50.5, 16.3], zoom_start=4)

budapest = [47.4850, 19.0548]
berlin = [52.4790, 13.3520]

# Degree markers
folium.Marker(
    budapest,
    popup="Bachelor's Degree at Corvinus University of Budapest",
    tooltip=folium.Tooltip("Corvinus University of Budapest", permanent=True, direction="top"),
    icon=folium.Icon(color="blue", icon="graduation-cap", prefix="fa")
).add_to(m)

folium.Marker(
    berlin,
    popup="Master's Degree at HWR Berlin",
    tooltip="Berlin",
    icon=folium.Icon(color="red", icon="graduation-cap", prefix="fa")
).add_to(m)

folium.PolyLine([budapest, berlin], color="purple", weight=3, opacity=0.8).add_to(m)

# Places featured in the journey carousel.
pecs = [46.076, 18.228]
pannonhalma = [47.553, 17.761]
for location, name, description, color, icon in [
    (pecs, "Pécs", "My hometown in southern Hungary.", "green", "home"),
    (pannonhalma, "Pannonhalma", "Where I went to high school.", "orange", "graduation-cap"),
]:
    folium.Marker(
        location,
        popup=f"<strong>{name}</strong><br>{description}",
        tooltip=folium.Tooltip(name, permanent=True, direction="top"),
        icon=folium.Icon(color=color, icon=icon, prefix="fa"),
    ).add_to(m)

m.fit_bounds([pecs, pannonhalma, budapest, berlin], padding=(50, 50))

# Support running from the repository root or the notebook folder.
site_dir = Path.cwd() / 'Intro Site' if (Path.cwd() / 'Intro Site').is_dir() else Path.cwd()
m.save(str(site_dir / 'my_map.html'))
```
