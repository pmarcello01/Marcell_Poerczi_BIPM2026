```python
import pandas as pd
df = pd.read_csv("weather_actual.csv")
print(df.head())
```

```python
df.columns=['time','date']
print(df.head())
```

```python
df["time"] = pd.to_datetime('')
print(df.head())
```

```python
df['year'] = df['time'].dt.year
df['month'] = df['time'].dt.month
df['day'] = df['time'].dt.day
```

```python
print(df.head())
```

```python
grouped_df = df.groupby(['year','month','day']).agg({
    "temperature_2m (°C)":"max",
    "relative_humidity_2m (%)":"mean",
    "precipitation (mm)":"mean",
    "pressure_msl (hPa)":"mean",
    }).reset_index()

novembers_df = grouped_df.loc[grouped_df['month']==10,:]
print(novembers_df)
```

```python
import numpy as np
novembers_df["week"] = np.ceil(novembers_df['day']/7)
## ceil = round_up / floor = round_down
print(novembers_df)
```

```python
avg_nov_temp = novembers_df["temperature_2m (°C)"].mean()
print(avg_nov_temp)
yearly_weekly_df = novembers_df.groupby(
    ["year", "week"]
).agg({
"temperature_2m (°C)":"mean"
})
yearly_weekly_df["vs_avg"]=avg_nov_temp-yearly_weekly_df["temperature_2m (°C)"]
yearly_weekly_df["vs_avg%"]=(avg_nov_temp-yearly_weekly_df["temperature_2m (°C)"])/avg_nov_temp

print(yearly_weekly_df)
```

```python
year_nov_df = yearly_weekly_df.groupby(['year']).agg({
    "temperature_2m (°C)":"mean",
    "vs_avg":"mean",
    "vs_avg%":"mean"
})
print(year_nov_df)
```

```python
week_nov_df = yearly_weekly_df.groupby(['week']).agg({
    "temperature_2m (°C)":"mean",
    "vs_avg":"mean",
    "vs_avg%":"mean"
})
print(week_nov_df)
```
