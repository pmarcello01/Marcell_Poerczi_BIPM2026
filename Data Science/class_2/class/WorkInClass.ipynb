Can we crate random numbers quickly ?

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

```

```python
# Uniform between 0 and 1
rng.random(10)
# default is a number between 0-1, argument here means size

# Uniform between 2 and 5
rng.uniform(2, 5, size=10)

# 10 random integers between 0 and 9
rng.integers(0, 10, size=10)
```

```python
# Standard normal: mean=0, standard deviation=1
rng = np.random.default_rng(42)

x = rng.normal(10, 5, size=200)
#this gives the values in normal distribution rng.normal(mean,stdev,size=...)
x = np.abs(x)**(2)
#print(x)
# Normal: mean=100, standard deviation=15
#rng.normal(100, 15, size=10)
xMean = x.mean()
print("mean:", xMean)
xMedian = np.median(x)
print("median:", xMedian)
```

### Mean Model


```python
residuals = x-xMean
L1mean = np.mean(np.abs(residuals))
#this is mean abs erros
print("L1 mean model", L1mean)
L2mean = np.sqrt(np.mean(np.abs(residuals)**2))
L2mean = (np.mean(np.abs(residuals)**2))**(1/2)
#this is Mean Squared Error

print("L2 mean model", L2mean)
```

### Median Model

```python
residuals = x-xMedian
L1median = np.mean(np.abs(residuals))#food for thought: we can also exted this to the median
print("L1 median model", L1median)
L2median = np.sqrt(np.mean(np.abs(residuals)**2))
print("L2 median model", L2median)

#Same absolute and squared errors, but measuring to the median
```

1. The mean minimizes L2
    The mean minimizes the sum of squared errors, and therefore also MSE and RMSE (your L2 calculation).
    - Sensitive to extreme values.
    - Useful when large errors should receive extra weight.
2. The median minimizes L1
    The median minimizes the sum of absolute errors, and therefore MAE (your L1 calculation).
    - Resistant to extreme values.
    - Often more representative of a typical observation in skewed data.
